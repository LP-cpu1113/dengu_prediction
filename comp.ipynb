```python
import pandas as pd
import numpy as np
```

### Loading the data...

```python
df_train = pd.read_csv("/home/anagha/dsr_group_project/dengu_prediction/data/dengue_features_train.csv")
df_train.head()
```

```python
df_train.shape
```

```python
df_labels = pd.read_csv("/home/anagha/dsr_group_project/dengu_prediction/data/dengue_labels_train.csv")
df_labels
```

```python
df_labels.shape
```

```python
df_merged = df_train.merge(df_labels, on=['city', 'year', 'weekofyear'])
df_merged['type'] = 'train'
df_merged
```

```python
df_test = pd.read_csv("data/dengue_features_test.csv")
df_test['type'] = 'test'
df_test
```

```python
df_final = pd.concat([df_merged, df_test], ignore_index=True)
df_final
```

```python
df_final.describe()
```

```python
df_final = df_final.drop('week_start_date', axis=1)
df_final.dtypes
```

#### Encoding categorical values

```python
df_final['city'] = df_final['city'].map({'sj': 0, 'iq': 1})
```

#### Replacing missing values with null

```python
cols_with_nulls = df_final.isnull().sum()[df_final.isnull().sum() > 0].index.tolist()
cols_with_nulls = [col for col in cols_with_nulls if col != 'total_cases']
#df_final[cols_with_nulls] = df_final[cols_with_nulls].fillna(df_final[cols_with_nulls].mean())
```

```python
train_df = df_final[df_final['type'] == 'train']
test_df  = df_final[df_final['type'] == 'test']

# Calculate mean from train only
train_mean = train_df[cols_with_nulls].mean()

# Fill both using train mean
train_df[cols_with_nulls] = train_df[cols_with_nulls].fillna(train_mean)
test_df[cols_with_nulls]  = test_df[cols_with_nulls].fillna(train_mean)

# Recombine
df_final = pd.concat([train_df, test_df], ignore_index=True)
```

```python
# Split by city
df_sj = df_final[df_final['city'] == 0]
df_iq = df_final[df_final['city'] == 1]

# Split by type for each city
train_sj = df_sj[df_sj['type'] == 'train']
test_sj  = df_sj[df_sj['type'] == 'test']

train_iq = df_iq[df_iq['type'] == 'train']
test_iq  = df_iq[df_iq['type'] == 'test']

# Features and target
features = [col for col in df_final.columns if col not in ['total_cases', 'type']]

X_train_sj = train_sj[features]
y_train_sj = train_sj['total_cases']

X_test_sj = test_sj[features]

X_train_iq = train_iq[features]
y_train_iq = train_iq['total_cases']

X_test_iq = test_iq[features]
```

```python
print("SJ train:", X_train_sj.shape, "SJ test:", X_test_sj.shape)
print("IQ train:", X_train_iq.shape, "IQ test:", X_test_iq.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

# Define models
rf_sj = RandomForestRegressor(n_estimators=100, random_state=42)
rf_iq = RandomForestRegressor(n_estimators=100, random_state=42)

# Fit
rf_sj.fit(X_train_sj, y_train_sj)
rf_iq.fit(X_train_iq, y_train_iq)

# Predict
pred_sj = rf_sj.predict(X_test_sj)
pred_iq = rf_iq.predict(X_test_iq)

# Round to nearest integer since cases are whole numbers
pred_sj = pred_sj.round().astype(int)
pred_iq = pred_iq.round().astype(int)

print("SJ predictions:", pred_sj[:5])
print("IQ predictions:", pred_iq[:5])
```

```python
# Create prediction dataframes
pred_df_sj = test_sj[['city', 'year', 'weekofyear']].copy()
pred_df_sj['total_cases'] = pred_sj

pred_df_iq = test_iq[['city', 'year', 'weekofyear']].copy()
pred_df_iq['total_cases'] = pred_iq

# Concatenate sj first then iq
preds = pd.concat([pred_df_sj, pred_df_iq], ignore_index=True)

preds
```

```python
preds['city'] = preds['city'].map({0: 'sj', 1: 'iq'})
preds
```

```python
preds.to_csv('predictions.csv', index=False)
```
